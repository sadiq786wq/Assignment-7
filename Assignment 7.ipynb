```python
class Employee:
    def __init__(self, name, number, address, salary, post, year_of_joining, education, years_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.years_of_experience = years_of_experience

    def describe_employee(self):
        print("Name:", self.name)
        print("Number:", self.number)
        print("Address:", self.address)
        print("Salary:", self.salary)
        print("Post:", self.post)
        print("Year of Joining:", self.year_of_joining)
        print("Education:", self.education)
        print("Years of Experience:", self.years_of_experience)

    def update_year_of_experience(self):
        self.years_of_experience = self.years_of_experience + 1
        print("Updated years of experience:", self.years_of_experience)

    def increase_salary(self, amount):
        self.salary = self.salary + amount
        print("New salary:", self.salary)
```

```python
employee101 = Employee("Ali Khan", "EMP101", "Karachi", 50000, "Data Analyst", 2020, "BS Computer Science", 3)
```

```python
employee101.describe_employee()
```

```python
employee101.increase_salary(5000)
```

```python
def make_employee(name, number, address, salary, post, year_of_joining, education, years_of_experience):
    employee = {"name": name, "number": number, "address": address, "salary": salary, "post": post, "year_of_joining": year_of_joining, "education": education, "years_of_experience": years_of_experience}
    return employee
```

```python
def describe_employee(employee):
    print("Name:", employee["name"])
    print("Number:", employee["number"])
    print("Address:", employee["address"])
    print("Salary:", employee["salary"])
    print("Post:", employee["post"])
    print("Year of Joining:", employee["year_of_joining"])
    print("Education:", employee["education"])
    print("Years of Experience:", employee["years_of_experience"])
```

```python
def update_year_of_experience(employee):
    employee["years_of_experience"] = employee["years_of_experience"] + 1
    print("Updated years of experience:", employee["years_of_experience"])
```

```python
def increase_salary(employee, amount):
    employee["salary"] = employee["salary"] + amount
    print("New salary:", employee["salary"])
```

```python
employee101 = make_employee("Ali Khan", "EMP101", "Karachi", 50000, "Data Analyst", 2020, "BS Computer Science", 3)
```

```python
print(employee101["name"])
```

```python
print(employee101["salary"])
```

```python
print(employee101["post"])
```
